## 👤 Step 2: Configure Your Profile & Upload Files

```python
#@title Candidate Information
first_name = "Meeran" #@param {type:"string"}
last_name = "Ahmed" #@param {type:"string"}
email = "meeranahmed17@gmail.com" #@param {type:"string"}
phone = "+91 8015186448" #@param {type:"string"}
city = "Vaniyambadi" #@param {type:"string"}
portfolio_url = "https://meeran-portfolio.antideploy.app" #@param {type:"string"}
gemini_api_key = "" #@param {type:"string"}

import os, yaml
os.makedirs('data_folder', exist_ok=True)

# Save Plain Text Resume YAML
profile_data = {
    'personal_information': {
        'name': first_name,
        'surname': last_name,
        'email': email,
        'phone': phone.replace('+91', '').strip(),
        'phone_prefix': '+91' if '+91' in phone else '',
        'city': city,
        'country': 'India',
        'portfolio': portfolio_url
    }
}
with open('data_folder/plain_text_resume.yaml', 'w') as f:
    yaml.dump(profile_data, f)

# Save Secrets
if gemini_api_key:
    with open('data_folder/secrets.yaml', 'w') as f:
        yaml.dump({'llm_api_key': gemini_api_key}, f)

print('✅ Profile & API Key configured!')

# Upload Resume & Jobs Spreadsheet
from google.colab import files
print('\n📄 Upload your Resume PDF (must be named resume.pdf or upload now):')
uploaded_resume = files.upload()
for fn in uploaded_resume.keys():
    os.rename(fn, 'data_folder/resume.pdf')
    print(f'Saved resume to data_folder/resume.pdf')

print('\n📊 (Optional) Upload your custom jobs.xlsx spreadsheet (or leave blank to use defaults):')
uploaded_jobs = files.upload()
for fn in uploaded_jobs.keys():
    os.rename(fn, 'data_folder/jobs.xlsx')
    print(f'Saved jobs spreadsheet to data_folder/jobs.xlsx')
```

## ⚡ Step 3: Run OmniSwarm Multi-Agent Engine

```python
# Launch autonomous browser swarm
!python swarm_main.py --mode excel --file data_folder/jobs.xlsx --limit 50

# Automatically download the executive report to your phone/tablet
from google.colab import files
import os

if os.path.exists('data_folder/output/APPLICATIONS_REPORT.md'):
    print('\n📥 Downloading your Applications Report...')
    files.download('data_folder/output/APPLICATIONS_REPORT.md')

if os.path.exists('data_folder/output/success.json'):
    files.download('data_folder/output/success.json')
print('\n🎉 All done! Check your downloads for the full ledger.')
```
